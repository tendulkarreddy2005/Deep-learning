Task 6: Bidirectional Long Short-Term Memory (BiLSTM) Cell
Mechanics from Scratch

```python
# Forward LSTM
i = sigmoid(Wi @ x + Ui @ h + bi)      # Input Gate
f = sigmoid(Wf @ x + Uf @ h + bf)      # Forget Gate
o = sigmoid(Wo @ x + Uo @ h + bo)      # Output Gate
g = np.tanh(Wc @ x + Uc @ h + bc)      # Cell Candidate
c = f * c + i * g                      # Cell State
h = o * np.tanh(c)                     # Hidden State

# Backward LSTM
h_back = h[::-1]                       # Reverse Sequence

# Bidirectional Output
output = np.concatenate((h, h_back), axis=1)
```
